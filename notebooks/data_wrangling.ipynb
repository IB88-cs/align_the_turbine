```python
#flagging and excluding 'stop' periods
stop_periods = operating_df[operating_df['Status'] == 'Stop']

intervals = pd.IntervalIndex.from_arrays(
    stop_periods['Timestamp start'],
    stop_periods['Timestamp end'],
    closed = 'left'
)

wind_df['during_stop'] = wind_df['Date and time'].apply(lambda t: intervals.contains(t).any())

merged_df = wind_df[~wind_df['during_stop']].drop(columns='during_stop')
```

```python
operating_df[operating_df['Status'] == 'Warning']['Message'].value_counts()
```

```python
# 1. Make sure timestamps are parsed (coerce handles the "-" placeholder rows)
operating_df['Timestamp start'] = pd.to_datetime(operating_df['Timestamp start'], errors='coerce')
operating_df['Timestamp end'] = pd.to_datetime(operating_df['Timestamp end'], errors='coerce')
wind_df['Date and time'] = pd.to_datetime(wind_df['Date and time'])

# 2. Full stop periods — turbine wasn't producing at all
stop_periods = operating_df[operating_df['Status'] == 'Stop']

# 3. Warnings that are specifically pitch- or wind-sensor related
exclude_keywords = ['Limit switch', 'Pitch measuring', 'anemometer', 'vane', 'Battery charge cycle']
pitch_related_warnings = operating_df[
    (operating_df['Status'] == 'Warning') &
    (operating_df['Message'].str.contains('|'.join(exclude_keywords), case=False, na=False))
]

# 4. Combine both into one exclusion set, drop rows with no valid end timestamp
exclude_periods = pd.concat([stop_periods, pitch_related_warnings]).dropna(subset=['Timestamp start', 'Timestamp end'])

# 5. Build an interval index and flag any wind_df row that falls inside an excluded period
intervals = pd.IntervalIndex.from_arrays(
    exclude_periods['Timestamp start'],
    exclude_periods['Timestamp end'],
    closed='left'
)
wind_df['excluded'] = wind_df['Date and time'].apply(lambda t: intervals.contains(t).any())

# 6. Final cleaned dataframe
merged_df = wind_df[~wind_df['excluded']].drop(columns='excluded')

print(f"Original rows: {len(wind_df)}")
print(f"Excluded rows: {wind_df['excluded'].sum()}")
print(f"Remaining rows: {len(merged_df)}")
```
